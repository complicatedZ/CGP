```python
import scipy.io as sio
import pickle
import matplotlib.pyplot as plt
import numpy as np
import sys
sys.path.append('../../')
from helper_function import *

plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 18
```

```python
cmap = plt.get_cmap('coolwarm')
num_colors = 20
colors = [cmap(i / num_colors) for i in range(num_colors)]

set_color = colors[19] #17 18
scatter_color = colors[18] #3 2

with open('../../random_walk/Ecoli_400_5_proteome_weakbg.pkl', 'rb') as pickle_file:
    data = pickle.load(pickle_file)
    all_total_nutrient_count = data['all_total_nutrient_count']
    all_total_intake = data['all_total_intake']
    all_growth_rate = data['all_growth_rate']
    all_random_medium = data['all_random_medium']
    all_met_flux = data['all_met_flux']
    del data
```

```python
set_panel_style(scale=4)

plt.figure(figsize=(8, 8))

for i in [3]:
    mask = all_growth_rate[i] > 0.01
    x = all_total_nutrient_count[i][mask]
    y = all_growth_rate[i][mask]

    if len(x) < 2:
        continue

    try:
        plt.scatter(
            x, y,
            color=scatter_color,
            label="Monocultures",
            s=200,
            edgecolors="none"
        )


    except Exception:
        continue

# Axes limits
plt.xlim((0, float(np.max(all_total_nutrient_count[i])) + 1))
plt.ylim((0, 5))

# Legend
plt.legend(frameon=False, handletextpad=0.2)

ax = plt.gca()


ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)



plt.tight_layout()
# plt.savefig("Figure2E.pdf", format="pdf", pad_inches=0.5)

plt.show()
```
