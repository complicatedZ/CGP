```python
# Data from Ho, PY., Nguyen, T.H., Sanchez, J.M. et al. Resource competition predicts assembly of gut bacterial communities in vitro. Nat Microbiol 9, 1036–1048 (2024).

data = np.array([
    [0.479, 3496],
    [0.645, 4341],
    [0.385, 2686],
    [0.097, 456],
    [0.376, 3636],
    [0.436, 3478],
    [0.214, 792],
    [0.360, 3684],
    [0.168, 3065],
    [0.470, 3758],
    [0.562, 2581],
    [0.531, 4540],
    [0.119, 1880],
    [0.092, 1016],
    [0.040, 323]
])
```

```python
import matplotlib.pyplot as plt
import numpy as np
set_panel_style(scale=4)

point_rgb = (0.0, 0.0, 0.0)   # gray
line_rgb  = (0.0, 0.0, 0.0)   # black
total_slope = []

x = data[:, 1]
y = data[:, 0]

plt.figure(figsize=(8, 8))

plt.scatter(
    x, y,
    color=point_rgb,
    label='Monocultures',
    alpha=1,
    s=180,
    edgecolors='none'      
)

# --- Fit line ---
slope, intercept = np.polyfit(x, y, 1)
xx = np.arange(0, int(np.ceil(np.max(x))) + 1)
yy = slope * xx + intercept
plt.plot(xx, yy, color=line_rgb, alpha=1)


print(pearson_ci(x, y, confidence=0.95))

# Axes limits
plt.xlim((0, int(np.ceil(np.max(x))) + 100))
plt.ylim((0, 0.8))

# --- Legend ---
plt.legend(frameon=False, handletextpad=0.05)

# Spines
ax = plt.gca()
ax.set_box_aspect(1)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.xlabel('#Metabolic features utilized')
plt.ylabel(r'Growth rate ($h^{-1}$)')
plt.tight_layout()
#plt.savefig('Figure2A.pdf', format="pdf", pad_inches=0.01)
plt.show()
```
